### Clean campaign_interactions


```python
# read data file

campaign_interactions = pd.read_csv(r"G:\courses\DEPI\TEQ\week5\Customer 360 project\Week 05\campaign_interactions.csv")
```

```python
#clean function
def clean_campaign_interactions(csv_file):

    # change coulmns name
    csv_file = csv_file.rename(columns = {"converted" : "campaign_converted"})

    # change columns name to lower casa and trim text
    csv_file.columns = csv_file.columns.str.strip().str.lower()

    # trim index of the columns
    for strip in ["interaction_type" , "campaign_id" , "interaction_id" , "campaign_converted"]:
        csv_file[strip] = csv_file[strip].str.strip()

    # change text to lower case in index of the columns
    for lower in ["campaign_converted" , "interaction_type"]:
        csv_file[lower] = csv_file[lower].str.lower()

    #  change str type to category
    for i in ['campaign_id', 'interaction_type', 'campaign_converted' , "interaction_id"] :
        csv_file[i] = csv_file[i].astype('category')

    # remove nan value
    csv_file.dropna(subset="campaign_id" , inplace= True)

    # change str type to date
    csv_file['interaction_date'] = pd.to_datetime(
    csv_file['interaction_date'],
    format='mixed',
    errors='coerce'
    )

    # remove dublicate rows
    csv_file = csv_file.drop_duplicates()

    return csv_file
```


### Clean subscriptions


```python
# read data file
subscriptions = pd.read_csv(r"G:\courses\DEPI\TEQ\week5\Customer 360 project\Week 05\subscriptions.csv")
```

```python
#clean function
def clean_subscriptions(csv_file):

    # rename some columns
    csv_file.rename(columns={"plan" : "sub_plan" , "billing_cycle" : "sub_billing_cycle", "monthly_fee" : "sub_monthly_fee", "discount" : "sub_discount", "status" : "sub_status" } , inplace= True)

    # remove start_date and end_date columns
    csv_file = csv_file.drop(columns= ["start_date" , "end_date"])

    # change columns name to lower casa and trim text
    csv_file.columns = csv_file.columns.str.strip().str.lower()

    # clean monthly_fee column
    csv_file["sub_monthly_fee"] = csv_file["sub_monthly_fee"].str.replace("USD" , "" , regex=False).str.replace("$" , "" , regex=False).str.replace("," , "" , regex=False).str.strip()
    csv_file["sub_monthly_fee"] = pd.to_numeric(csv_file["sub_monthly_fee"] , errors="coerce")

    # fill nall values in monthly_fee
    csv_file["sub_monthly_fee"] = csv_file["sub_monthly_fee"].fillna(csv_file["sub_monthly_fee"].median())

    # clean discount column
    csv_file["sub_discount"] = csv_file["sub_discount"].str.replace(" ","").str.replace("%" , "")
    csv_file["sub_discount"] = (csv_file["sub_discount"].fillna(0).astype("float")) / 100

    # trim index of the columns
    for strip in ["subscription_id", "sub_plan", "sub_billing_cycle", "sub_status"]:
        csv_file[strip] = csv_file[strip].str.strip()

    # change text to lower case in index of the columns
    for lower in ["sub_plan" , "sub_billing_cycle" , "sub_status"]:
        csv_file[lower] = csv_file[lower].str.lower()

    # change str type to category
    for i in ["sub_plan" , "sub_billing_cycle" , "sub_status" , "subscription_id"] :
        csv_file[i] = csv_file[i].astype('category')
    
    # remove dublicate rows
    csv_file = csv_file.drop_duplicates()
    

    return csv_file
```


### Clean Customers


```python
# read data file

customers = pd.read_csv(r"G:\courses\DEPI\TEQ\week5\Customer 360 project\Week 05\customers.csv")
```

```python
#clean function
def clean_customers(csv_file):

    # change str type to date
    for date_col in ["birth_date" , "registration_date" , "last_login_date"]:
        csv_file[date_col] = pd.to_datetime(csv_file[date_col], format='mixed', errors='coerce')

    # remove some coulmns
    csv_file = csv_file.drop(columns = ["email_address", "phone_number" , "address"])

    # remove dublicate rows by user id
    csv_file.drop_duplicates(subset= "user_id" , inplace=True)

    # change columns name to lower casa and trim text
    csv_file.columns = csv_file.columns.str.strip().str.lower()

    # clean annual_revenue column and fill nan values
    csv_file["annual_revenue"] = csv_file["annual_revenue"].str.replace("USD" , "" , regex=False).str.replace("$" , "" , regex=False).str.replace("," , "" , regex=False).str.strip()
    csv_file["annual_revenue"] = pd.to_numeric(csv_file["annual_revenue"] , errors="coerce")

    # fill nan values in "annual_revenue" coulmn
    csv_file["annual_revenue"] = csv_file["annual_revenue"].fillna(csv_file["annual_revenue"].median())
    
    # fill nan values in company_name coulmn
    csv_file["company_name"] = csv_file["company_name"].fillna("unknown")

    # fill nan values in last_login_date coulmn
    csv_file["last_login_date"] = csv_file["last_login_date"].ffill()
    
    # trim index of the columns
    for strip in ["full_name", "job_title", "company_name", "city", "country", "account_status"]:
        csv_file[strip] = csv_file[strip].str.strip()

    # change text to lower case in index of the columns
    for lower in ["full_name", "job_title", "company_name", "city", "country", "account_status"]:
        csv_file[lower] = csv_file[lower].str.lower()

    # change str type to category
    for i in ["full_name", "job_title", "company_name", "city", "country", "account_status"] :
        csv_file[i] = csv_file[i].astype('category')

    return csv_file
```
